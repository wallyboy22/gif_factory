# 🏭 Fábrica de GIFs — MapBiomas

Pipeline para gerar GIFs animados de dados ambientais do **MapBiomas** via
[Google Earth Engine](https://earthengine.google.com/) e publicar no
[Looker Studio](https://lookerstudio.google.com/).

---

## 📖 Como usar

1. **Célula 1** (Setup) — clone do repositório, instala dependências, autentica GEE e GCS
2. Escolha uma **seção** (Fogo ou Degradação) e execute as células desejadas
3. Cada célula é **independente** — pode pular ou reordenar
4. Se o Colab **desconectar**, reexecute a célula com `--resume` que ele retoma de onde parou
5. Ao final, execute a célula **Sync** para subir tudo ao GCS e atualizar o Looker Studio

> ⚡ Recomendado: **Colab Pro** (mais RAM, sessão mais longa, 8 workers)
>
> 🔁 Workers sugeridos: 8 para Fogo, 6 para Degradação
