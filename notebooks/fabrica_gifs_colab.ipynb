# Fábrica de GIFs — piloto

Navegue pelo catálogo existente, busque territórios em uma FeatureCollection e gere GIFs. Colagens são opcionais; frames intermediários são preservados.

O inventário é apenas para consulta. Pastas existentes são puladas, sem verificar equivalência. É necessário acesso aos assets e uma cópia atualizada do repositório.

```python
# A pasta existente nunca é apagada ou atualizada automaticamente.
REPOSITORIO = '/content/gif_factory'
PROJETO_EARTH_ENGINE = 'SEU-PROJETO-GOOGLE-CLOUD'
PASTA_SAIDA = '/content/gif_factory_outputs'

from pathlib import Path
import os, sys, subprocess
raiz = Path(REPOSITORIO).resolve()
if not raiz.exists():
    subprocess.run(['git', 'clone', 'https://github.com/wallyboy22/gif_factory.git', str(raiz)], check=True)
if not (raiz / 'src/mapbiomas_data/interfaces/colab_gifs.py').is_file():
    raise FileNotFoundError('A pasta existente não contém a versão atualizada; ajuste REPOSITORIO.')
os.chdir(raiz)
sys.path.insert(0, str(raiz / 'src'))
```
